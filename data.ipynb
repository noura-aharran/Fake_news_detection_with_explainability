```python
from collections import Counter
dataset = load_dataset("davanstrien/WELFake")
labels = [row['label'] for row in dataset['train']]
print(Counter(labels))
```

```python
ex = dataset['train'][0]
print("Titre :", ex["title"])
print("Texte :", ex["text"])
print("Label :", "Fake" if ex["label"]==0 else "Real")
```

```python
import matplotlib.pyplot as plt
from collections import Counter

labels = [row['label'] for row in dataset['train']]
counter = Counter(labels)

plt.bar(["Fake", "Real"], [counter[0], counter[1]])
plt.title("Répartition des articles")
plt.show()
```
